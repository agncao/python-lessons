## id() 函数的官方定义
它是 [Python 语言]((https://docs.python.org/3/reference/datamodel.html))的标准特性，根据 Python 官方规范，id() 是一个内置函数，其定义如下：
- 必须返回一个整数，用来代表该对象的“唯一身份（Identity）”。
- 该整数在对象的生命周期内必须是唯一且恒定不变的。两个同时存在的不同对象，其 id() 绝对不能相同。
- 只要写 Python 代码，无论在哪个平台、哪个解释器上，都可以调用 id(a)。

## 解释器关于此函数的特有实现
- **CPython (官方默认)** 对象的内存物理地址（对应 C 语言的指针）。
- **PyPy(追求高性能)** 一个递增的唯一整数计数器。因为 PyPy 会在内存中移动对象以优化性能，所以它绝对不能直接用内存地址当做固定的 ID。
- 如果一个对象被销毁（垃圾回收）了，它的 id 可能会被分配给之后创建的新对象。
