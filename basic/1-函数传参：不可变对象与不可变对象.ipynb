Python 的函数参数传递本质是“按对象引用传递”（pass by object reference），也就是将对象的内存地址传递给函数。
## 核心机制：传对象引用
- Python 中的变量只是一个贴在对象上的“标签”。
- 函数传参时，实参将引用（地址）赋值给形参，相当于函数内部多了一个指向同一个对象的标签。
- 传入后表现如何，取决于对象是可变对象还是不可变对象。

## 不可变对象（如 int, str, tuple）
- 传递不可变对象时，在函数内修改变量值，会让形参指向一个新的对象。
- 原外的实参标签不受影响，值保持不变。
- 示例：

```python
def change(x):
    x = 2  # x 指向了新对象 2
a = 1
change(a)
print(a)  # 输出 1，a 没变
```

## 可变对象（如 list, dict, set）
- 传递可变对象时，在函数内通过方法（如 append、update）修改对象内容，原对象直接改变。
- 外部的实参也会同步看到修改后的内容。
- 示例：

```python
def append_item(lst):
    lst.append(1)  # 直接修改原列表对象
a = [0]
append_item(a)
print(a)  # 输出 [0, 1]，a 变了
```
